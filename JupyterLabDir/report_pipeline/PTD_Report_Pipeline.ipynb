# PTD Report Pipeline
This notebook is a thin wrapper over the report pipeline CLI scripts.


```python
from pathlib import Path
import subprocess
import pandas as pd
from IPython.display import Image, display

REPO_ROOT = Path('/home/admin/workspace/NErlNet')
PIPELINE_DIR = REPO_ROOT / 'JupyterLabDir' / 'report_pipeline'
MANIFEST = PIPELINE_DIR / 'device_manifest.json'
MATRIX = PIPELINE_DIR / 'experiment_matrix.report_core.json'
JSON_OUT = PIPELINE_DIR / 'generated_jsons'
RUN_OUT = PIPELINE_DIR / 'output_report_core'
QUICK_OUT = PIPELINE_DIR / 'output_quick'

def run_cmd(cmd):
    print(' '.join(str(x) for x in cmd))
    subprocess.run([str(x) for x in cmd], check=True)
```

```python
run_cmd([
    'python3', PIPELINE_DIR / 'generate_jsons.py',
    '--profile', 'report_core',
    '--manifest', MANIFEST,
    '--matrix', MATRIX,
    '--out', JSON_OUT,
])
```

```python
run_cmd([
    'python3', PIPELINE_DIR / 'run_matrix.py',
    '--profile', 'report_core',
    '--manifest', MANIFEST,
    '--matrix', MATRIX,
    '--json-dir', JSON_OUT,
    '--out', QUICK_OUT,
    '--quick',
    '--fail-fast',
])
```

```python
phase_summary = pd.read_csv(QUICK_OUT / 'raw' / 'phase_summary.csv')
run_summary = pd.read_csv(QUICK_OUT / 'raw' / 'run_summary.csv')
display(phase_summary)
display(run_summary)
```
